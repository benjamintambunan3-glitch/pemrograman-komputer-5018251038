```python

sensor = [25.5, 30.2, -5.0, 45.0, 28.5,
          60.0, 32.0, 15.5, 75.0, 40.0,
          22.0, 55.5, 18.0, 90.0, 35.0,
          27.5, -2.0, 48.0, 65.0, 33.5]

total = 0
jumlah_valid = 0
jumlah_alarm = 0
minimum = 0
maksimum = 0

print("=== SENSOR LOG ANALYZER ===")

# Menampilkan nomor sampel dan nilai sensor
for i in range(len(sensor)):
    print("Sampel", i + 1, "=", sensor[i])

# Mengolah data sensor
for i in range(len(sensor)):
    nilai = sensor[i]

    # Mengabaikan nilai negatif
    if nilai < 0:
        print("Sampel", i + 1, ": DATA TIDAK VALID")
    else:
        total += nilai
        jumlah_valid += 1

        # Menentukan status NORMAL/ALARM
        if nilai > 50:
            print("Sampel", i + 1, ": ALARM")
            jumlah_alarm += 1
        else:
            print("Sampel", i + 1, ": NORMAL")

        # Mencari minimum dan maksimum secara manual
        if jumlah_valid == 1:
            minimum = nilai
            maksimum = nilai
        else:
            if nilai < minimum:
                minimum = nilai

            if nilai > maksimum:
                maksimum = nilai

# Menghitung rata-rata
if jumlah_valid > 0:
    rata_rata = total / jumlah_valid

    print("\n=== HASIL ANALISIS ===")
    print("Jumlah data valid =", jumlah_valid)
    print("Jumlah alarm =", jumlah_alarm)
    print("Rata-rata =", rata_rata)
    print("Nilai minimum =", minimum)
    print("Nilai maksimum =", maksimum)
else:
    print("Tidak ada data sensor yang valid.")
```
