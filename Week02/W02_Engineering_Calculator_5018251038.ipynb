```python
voltage = float(input("Voltage [V]: "))
current = float(input("Current [A]: "))
operating_time = float(input("Operating Time [hour]: "))

power = voltage * current
energy = power * operating_time

print("===== ELECTRICAL CALCULATION =====")
print(f"Voltage = {voltage:.2f} V")
print(f"Current = {current:.2f} A")
print(f"Power = {power:.2f} W")
print(f"Operating Time = {operating_time:.2f} hour")
print(f"Energy = {energy:.2f} Wh")
```

# Jawaban Refleksi

1. int digunakan untuk menyimpan bilangan bulat tanpa angka desimal, sedangkan float digunakan untuk menyimpan bilangan yang memiliki angka desimal.
2. Karena secara default input() menghasilkan data bertipe string (teks). Agar data tersebut dapat digunakan dalam operasi perhitungan numerik, data perlu dikonversi menjadi float.
3. Operator = digunakan untuk memberikan atau menyimpan nilai ke dalam variabel, sedangkan == digunakan untuk membandingkan apakah dua nilai sama. Hasil perbandingan == berupa True atau False.
4. and digunakan untuk menggabungkan dua kondisi dan menghasilkan True jika kedua kondisi benar. Sedangkan or menghasilkan True jika minimal salah satu kondisi benar.
5. Kesalahan yang saya temui adalah sempat lupa menggunakan float() pada input. Akibatnya, nilai yang dimasukkan terbaca sebagai teks dan menyebabkan masalah saat melakukan perhitungan. Saya memperbaikinya dengan mengubah input menjadi float sebelum digunakan dalam operasi matematika.
